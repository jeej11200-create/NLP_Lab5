# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sentences)

similarity_matrix = cosine_similarity(tfidf_matrix)

df_similarity = pd.DataFrame(
    similarity_matrix,
    index=[f"Sentence {i+1}" for i in range(len(sentences))],
    columns=[f"Sentence {i+1}" for i in range(len(sentences))]
)

df_similarity
````

**Conclusion:**
- Sentences 1 and 4 have a similarity of **1.0** even though their word order differs. TF-IDF uses a Bag-of-Words representation, which ignores word order and punctuation, so both sentences produce identical vectors.
- Sentence 2 is fairly similar to sentences 1 and 4 (**0.65**) because they share words like *document*, *this*, *is*, and *the*.
- Sentence 3 is the least similar to the others (**0.23 – 0.31**) because it shares only common words (*this, is, the*) and contains unique words (*and, third, one*).

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

sentences = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sentences)

df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

df_tfidf

````

````python
for i, row in df_tfidf.iterrows():
    top = row.sort_values(ascending=False).head(3)
    print(f"Sentence {i+1}:", list(top.index))
````

**Important words:**
- **Sentence 1:** *science* has the highest score because it appears twice. *of* also scores high, which shows a limitation of TF-IDF without stopword removal. *fields*, *important*, and *most* tie with the same score since they are unique to this sentence.
- **Sentence 2:** *best*, *courses*, and *this* are the most important because they appear only in this sentence.
- **Sentence 3:** *data* has the highest score (0.64) because it appears twice in a short sentence, followed by *analyze* and *scientists*, which are unique to this sentence.

Words that appear in many sentences (e.g. *data*, *is*, *the*) receive lower IDF weights, while rare or repeated words in a single sentence receive higher scores.

````python
import pandas as pd
import re

df_simpsons = pd.read_csv(
    '/content/simpsons_script_lines.csv',
    low_memory=False
)

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

df_simpsons = df_simpsons.dropna(subset=['spoken_words'])
df_simpsons['cleaned_text'] = df_simpsons['spoken_words'].apply(clean_text)

df_simpsons[['spoken_words', 'cleaned_text']].head()

````

````python
from nltk.tokenize import word_tokenize
import nltk
import gensim

nltk.download('punkt')
nltk.download('punkt_tab')

tokens = [
    word_tokenize(text)
    for text in df_simpsons['cleaned_text']
]

Skip_gram_model = gensim.models.Word2Vec(
    tokens,
    min_count=1,
    vector_size=100,
    window=5,
    sg=1
)

print("Skip-Gram model trained successfully!")
print("Vocabulary size:", len(Skip_gram_model.wv))
````

**Conclusion:**
The `spoken_words` column was cleaned by removing missing values, converting text to lowercase, and removing digits and punctuation. The cleaned text was then tokenized with NLTK and used to train a Skip-gram Word2Vec model (`sg=1`, `vector_size=100`, `window=5`). The resulting vocabulary contains **44,282** unique tokens.

Note: the given cleaning function does not remove `?` and `!`, so they remain as separate tokens.

````python
print("Words most similar to homer:")
print(Skip_gram_model.wv.most_similar('homer'))

print("\nWords most similar to marge:")
print(Skip_gram_model.wv.most_similar('marge'))

print("\nWords most similar to bart:")
print(Skip_gram_model.wv.most_similar('bart'))
````
